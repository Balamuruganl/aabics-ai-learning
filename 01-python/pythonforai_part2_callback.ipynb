```python
def add(a,b):
    c=a+b
    print(f"The sum of {a} and {b} is {c}")
```

```python
def sub(a,b):
    c=a-b
    print(f"The difference of {a} and {b} is {c}")
```

```python
def multiply(a,b):
    c=a*b
    print(f"The product of {a} and {b} is {c}") 
```

```python
add(20,30)
sub(50,20)
multiply(10,20)
```

```python
def calculate(a,b,operation):
    operation(a,b)
```

```python
calculate(10,20,add)
calculate(50,20,sub)
calculate(10,20,multiply)
```
