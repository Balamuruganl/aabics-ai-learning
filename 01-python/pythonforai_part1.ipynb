```python
#tuple-immutable
words = ("hi","hello","ai",123)
words[0]
```

```python
#list-mutable
words = ["hi","hello","ai","123"]
words[2] = "machine learning"
words
#list of tuples
messages = [("human","I want to learnAI"),
             ("AI","Thats a great idea") ]
messages[0]
```

```python
#list-comprehension
sentences = ["How do I learn AI?",
             "explain machine learning",
             "hi",
             "AI"]

result =[]
#traditional for loop
for sentence in sentences:
    if len(sentence)>5:
        result.append(sentence)
print(result)
#[expression iteration condition]
result_comprehension =[sentence for sentence in sentences if len(sentence)>10]
result_comprehension
```

```python
#Dictionary-mutable
student={"name":"Bala",
         "age":25,
         "subject":"ai"}
print(student["age"])
student["age"]=30
student
```

```python
#list of dictionaries
students = [{"name":"Bala","age":25,"subject":"ai"},
            {"name":"Ravi","age":30,"subject":"ml"}]
students[0]["name"]
```

```python
#nested dictionary
students = {"Bala":{"age":25,"subject":"ai"},
            "Ravi":{"age":30,"subject":"ml"},
            "Priya":{"age":28,"subject":"ds"},
            "Suresh":{"age":35,"subject":"ds"}}

students["Bala"]["age"]
```

```python
for key,value in students.items():
    print(key)
    print(value["age"])
```

```python
#dictionary comprehension
result = {key:value for key,value in students.items() if value["subject"]=="ds"}
result
```
